```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np

X, y, coef = make_regression(n_samples=100,
                         n_features=1,
                         n_informative=1,
                         noise=10,
                         coef=True,
                         random_state=0,
                         bias=100.0)
plt.scatter(X, y)
plt.title("X (1 dimension) and y")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

```python
model = LinearRegression()
model.fit(X, y)

print(f"Q2: y = {model.coef_[0]} * x + {model.intercept_}")

plt.scatter(X, y)
plt.title("X (1 dimension) and y")
plt.xlabel("X")
plt.ylabel("y")
plt.plot(X, model.predict(X), color='red')
plt.show()
```

```python
model.predict(X)[:10]
```

```python
def compute_mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

y_pred = model.predict(X)

mse = compute_mse(y, y_pred)

print(mse)
```

```python
X50, y50 = make_regression(n_samples=100,
                         n_features=1,
                         n_informative=1,
                         noise=50,
                         random_state=0,
                         bias=100.0)

model.fit(X50, y50)
y_pred50 = model.predict(X50)
mse50 = compute_mse(y50, y_pred50)

print(mse50)
```
