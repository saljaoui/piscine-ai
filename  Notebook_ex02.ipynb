```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def sigmoid_base(x):
    return sigmoid(x)

def sigmoid1(x):
    return sigmoid(0.5 * x + 3)

def sigmoid2(x):
    return sigmoid(5 * x + 11)

x = np.linspace(-20, 15, 100)

y_base = sigmoid_base(x)
y1 = sigmoid1(x)
y2 = sigmoid2(x)


plt.plot(x, y_base)
plt.plot(x, y1,)
plt.plot(x, y2,)

plt.plot([-20.0, 15.0], [0.5, 0.5], color="red")

plt.xlabel("x")
plt.title("Sigmoid")
plt.show()
```
