```python
from sklearn.linear_model import LinearRegression

X, y = [[1],[2.1],[3]], [[1],[2],[3]]

model = LinearRegression()

model.fit(X, y)

x_pred = [[4]]
y_pred = model.predict(x_pred)

y_pred
```

```python
print("Coefficient (a):", model.coef_)
print("Intercept (b):", model.intercept_)
print("Score (R2):", model.score(X, y))
```
