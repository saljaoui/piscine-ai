```python
data = [
    (12, 15, 16, "Bob"),
    (10, 9, 10, "Eli"),
    (18, 18, 19, "Tom"),
    (13, 14, 16, "Ryan")
]

neuron_h1 = Neuron(0.05, 0.001, 0, regression=False)
neuron_h2 = Neuron(0.002, 0.003, 0, regression=False)
neuron_o1 = Neuron(2, 7, 10, regression=True)

targets = []
predictions = []

for math_g, chem_g, phys_g, name in data:
    h1_out = neuron_h1.feedforward(math_g, chem_g)
    h2_out = neuron_h2.feedforward(math_g, chem_g)

    y_hat = neuron_o1.feedforward(h1_out, h2_out)

    print(f"{name}: {y_hat}")
    targets.append(phys_g)
    predictions.append(y_hat)

# print("Targets:", targets)
# print("Predictions:", predictions)
mse = sum((y - y_hat) ** 2 for y, y_hat in zip(targets, predictions)) / len(targets)
print("MSE:", mse)
```
